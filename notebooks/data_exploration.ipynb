```python
plt.figure(figsize=(10, 6))
plt.scatter(df['Order Item Discount'],df['Order Profit Per Order'] , alpha=0.5)
plt.title('Product Price vs Order Profit Per Order')
plt.xlabel('Order Item Discount')
plt.ylabel('Order Profit Per Order')
plt.show()  
```

```python
#orders with losses
df[df['Order Profit Per Order']<0]['Order Profit Per Order']
```

```python
#most products causing loss when discounts are zero
top_loss_products = df[(df['Order Profit Per Order']<0) & (df['Order Item Discount']==0)].groupby('Product Name').agg(
    count=('Order Profit Per Order', 'count'),
    total_loss=('Order Profit Per Order', 'sum')
).sort_values(by='total_loss', ascending=True).head(10)

print(top_loss_products)
```

```python
#mean profit and loss percentage by delivery status
delivery_profit = df.groupby('Delivery Status').agg(
    mean_profit=('Order Profit Per Order', 'mean'),
    loss_percent=('Order Profit Per Order', lambda x: (x < 0).sum() / len(x) * 100)
)
print(delivery_profit)
```

```python
# بنشوف أنهي نوع شحن بيعمل أعلى نسبة خسائر في الشركة
shipping_loss = df.groupby('Shipping Mode').agg(
    mean_discount=('Order Item Discount', 'mean'),
    mean_profit=('Order Profit Per Order', 'mean'),
    count_of_loss_orders=('Order Profit Per Order', lambda x: (x < 0).sum())
).sort_values(by='mean_profit')
print(shipping_loss)
```

```python
#the five highest states in sales and profit
state_analysis = df.groupby('Customer State').agg(
    total_sales=('Sales', 'sum'),
    total_profit=('Order Profit Per Order', 'sum')
).sort_values(by='total_profit', ascending=False)
print(state_analysis.head(5)) # the highest in profit
print(state_analysis.tail(5)) # the highest in loss
```
